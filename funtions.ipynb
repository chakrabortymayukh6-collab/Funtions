```python
def sum_even(numbers):
  total = 0
  for num in numbers:
    if num % 2 == 0:
      total += num
      return total
```

```python
print(sum_even([1,2,3,4,5]))
```

```python
def reverse_string(text):
    return text[::-1]

print(reverse_string("Python"))
```

```python
def square_numbers(numbers):
    return [n ** 2 for n in numbers]

print(square_numbers([1, 2, 3, 4, 5]))
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

for num in range(1, 201):
    if is_prime(num):
        print(num, end=" ")
```

```python
class Fibonacci:
    def __init__(self, terms):
        self.terms = terms
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.terms:
            raise StopIteration

        result = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return result


fib = Fibonacci(10)

for num in fib:
    print(num, end=" ")
```

```python
def powers_of_two(max_exp):
    for exp in range(max_exp + 1):
        yield 2 ** exp

for power in powers_of_two(5):
    print(power, end=" ")
```
